```python
#conjunto de dados e pre-processamento
import numpy as np
import matplotlib.pyplot as plt 


#recurso para formatar imagens
def format_image(image, label):
    """Redimenciona a img para 224x224 e normaliza os pixels entre 0 e 1
        Args: image: tensor da imagem original
        label: etiqueta associada a imagem
        Devoluções: imagem redimensionada e normalizada, etiqueta original
    """
    image = tf.image.resize(image, (224, 224))/255.0 #redimensiona e normaliza
    return image, label


#carregar conjuntos de dados 'cats_vs_dogs' dos conjunto de dados do tensorflow
#dividimos o conjunto de dados em 3 subconjuntos: treino, validação e teste
(raw_train, raw_validation, raw_test), metadata = tfds.load(
    'cats_vs_dogs',                         # Nome del dataset
    split=['train[:80%]', 'train[80%:90%]', 'train[90%:]'],  # Suddivisione personalizzata
    with_info=True,                         # Carica anche informazioni aggiuntive sul dataset
    as_supervised=True,                     # Ritorna coppie (immagine, etichetta)
)

#mostramos algumas informações sobre o conjunto de dados
num_examples = metadata.splits['train'].num_examples #numero total de exemplos  no conjunto de dados do treinamento
num_classes = metadata.features['label'].num_classes #numero de classes no conjunto de dados(no caso, 2: gato e cachorro)
print(num_examples)
print(num_classes)
BATCH_SIZE = 32
train_batches = raw_train.shuffle(num_examples//4).map(format_image).batch(BATCH_SIZE).prefetch(1)
#criamos lotes para validação: formatamos e usamos lotes de tamanho definido
validation_batches = raw_validation.map(format_image).batch(BATCH_SIZE).prefetch(1)
#criamos lotes para teste: formatamos imagem e usamos lotes de tamanho 1
test_batches = raw_test.map(format_image).batch(1)
#exibição das formas de dados
#pegamos um lote do conjunto de dados de treinamento para verificar o tamanho das imagens e rotulos
for image_batch, label_batch in train_batches.take(1): #extrair um lote do conjunto de dados de treinamento
    pass
image_batch.shape #mostramos a forma do lote de imagens
```
